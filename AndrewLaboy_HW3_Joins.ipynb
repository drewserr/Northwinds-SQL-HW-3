Propostion 1: Im going to be using INNER JOIN to generate a departments list of successful sales. By linking the Orders table with the Customers Table, I can show which comapny placed a specific order.

```sql
SELECT 
    O.OrderID, 
    C.CustomerCompanyName AS CustomerName, 
    O.OrderDate
FROM Sales.[Order] AS O
INNER JOIN Sales.Customer AS C 
    ON O.CustomerID = C.CustomerID;
```

Propostion 2: In this proposition we want to find customers who are registered into the system but never placed an order. We can use LEFT OUTER JOIN to see each customer who matches this and filter those that have a ORDER ID of NULL, meaning zero sales.

```sql
SELECT
    C.CustomerID,
    C.CustomerCompanyName
    FROM sales.customer AS C
    LEFT OUTER JOIN Sales.[ORDER] AS O
     ON C.CustomerID = O.CustomerID
     WHERE O.OrderID IS NULL;
```

Propostion 3: For this proposition an HR team would need to know who reports to who. Someone who is the Boss and someone who is a normal staff member are both employeess in the same list. So we need to link the list to itself to show the relationship between those employees.

```sql
SELECT 
    E.EmployeeFirstName + ' ' + E.EmployeeLastName AS StaffMember,
    M.EmployeeFirstName + ' ' + M.EmployeeLastName AS ManagerName
FROM HumanResources.Employee AS E
LEFT OUTER JOIN HumanResources.Employee AS M 
    ON E.EmployeeManagerId = M.EmployeeId;
```
